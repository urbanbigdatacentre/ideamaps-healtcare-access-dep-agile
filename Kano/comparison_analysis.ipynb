# Results Analysis and Visualization
## E2SFCA Accessibility Analysis - Kano State, Nigeria

This notebook analyzes and visualizes the results from E2SFCA analysis.

```python
# Import libraries
import sys
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import geopandas as gpd
%pip install matplotlib seaborn

import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
# Set paths to access Kano data
# Define directories
data_inputs = '../Kano/Data/raw/'
data_temp = '../Kano/Data/processed/'
model_outputs = '../Kano/Data/outputs/'
```

## 1. Load E2SFCA Results

Load the accessibility scores and classifications from the E2SFCA analysis.

```python
E2SFCA_results = gpd.read_file(data_temp + 'accessibility-index-classified.gpkg', layer='accessibility-index-classified')
```

```python
E2SFCA_results['duration_minutes'] = E2SFCA_results['duration_seconds'] / 60

result_map = {0: 'Low', 1: 'Medium', 2: 'High'}
E2SFCA_results['result'] = E2SFCA_results['result'].map(result_map)

E2SFCA_results['result'] = pd.Categorical(
    E2SFCA_results['result'],
    categories=['Low', 'Medium', 'High'],
    ordered=True
)

summary = E2SFCA_results.groupby(['Local_Validation', 'result']).agg(
    Avg_Duration_Min=('duration_minutes', 'mean'),
    Avg_Distance_KM=('distance_km', 'mean')
).round(2)

print(summary)
```
